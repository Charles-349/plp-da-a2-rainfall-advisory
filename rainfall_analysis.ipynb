## Introduction

This project analyses monthly rainfall data collected from five Kenyan counties: Nairobi, Nakuru, Kisumu, Mombasa and Machakos.

The objective is to clean the rainfall data, analyse annual and monthly rainfall patterns using NumPy, identify wettest months and dry months, and recommend the most suitable consecutive two-month planting windows for each county.

The analysis uses Python, NumPy and the CSV module to produce a rainfall summary and practical planting advisory for a farmer-support NGO.

```python
import numpy as np
import csv
```

```python
# Import the required libraries

import numpy as np
import csv

print("Libraries imported successfully.")
```

```python
counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]

clean_rows = [

    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],  # Nairobi

    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],   # Nakuru

    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],  # Kisumu

    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70],  # Mombasa

]

machakos_raw = ["48", "35", "88", "N/A", "140", " 20 ",

                "9", "12", "", "45", "182", "96"]
```

```python
# Define the twelve months

months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]
```

## Task 1: Machakos Data Cleaning

In this task, I clean the raw Machakos rainfall data by converting valid text values into integers. I use exception handling to identify invalid entries and replace them with the rounded average of the valid rainfall readings. I also record and display the months where missing or invalid values were imputed.

```python
def clean_machakos(raw_data, month_names):
    """
    Convert Machakos rainfall readings into integers.
    Replace invalid readings with the rounded average
    of valid readings and return the cleaned data.
    """

    cleaned_values = []
    imputed_months = []

    # Convert raw text values into integers

    for month, value in zip(month_names, raw_data):

        try:
            cleaned_values.append(int(value.strip()))

        except ValueError:
            cleaned_values.append(None)
            imputed_months.append(month)

    # Calculate the average of valid readings

    valid_values = [value for value in cleaned_values if value is not None]

    average_rainfall = round(sum(valid_values) / len(valid_values))

    # Replace invalid values with the rounded average

    cleaned_values = [
        average_rainfall if value is None else value
        for value in cleaned_values
    ]

    return cleaned_values, imputed_months, average_rainfall


# Clean Machakos rainfall data

machakos_clean, imputed_months, imputed_value = clean_machakos(
    machakos_raw, months
)


# Display the results

print("MACHAKOS DATA CLEANING REPORT")
print("-" * 40)

print("Cleaned Rainfall Values:", machakos_clean)

print("Valid Data Average:", imputed_value, "mm")

print("Imputed Months:", ", ".join(imputed_months))

print("Number of Imputed Values:", len(imputed_months))
```

## Task 2: Creating the Combined NumPy Array

In this task, I combine the four existing county datasets with the cleaned Machakos data into one NumPy array. I then inspect its dimensions and data type to confirm that the dataset contains five counties and twelve monthly rainfall readings.

```python
# Combine all five counties into one NumPy array

R = np.array(clean_rows + [machakos_clean], dtype=int)


# Display the combined array

print("COMBINED RAINFALL ARRAY")
print("-" * 40)

print(R)


# Confirm the array shape and data type

print("\nArray Shape:", R.shape)

print("Array Data Type:", R.dtype)
```

## Task 3: Annual Rainfall Totals and Monthly Averages

In this task, I use NumPy axis operations to calculate annual rainfall totals for each county and average rainfall for each month across all five counties. These calculations help identify differences in annual rainfall and seasonal rainfall patterns.

```python
# Calculate annual rainfall totals for each county

annual_totals = np.sum(R, axis=1)


# Calculate monthly average rainfall across all counties

monthly_averages = np.mean(R, axis=0)


# Display annual rainfall totals

print("ANNUAL RAINFALL TOTALS")
print("-" * 40)

for county, total in zip(counties, annual_totals):

    print(f"{county:<12}: {total} mm")


# Display monthly rainfall averages

print("\nMONTHLY AVERAGE RAINFALL")
print("-" * 40)

for month, average in zip(months, monthly_averages):

    print(f"{month:<12}: {average:.2f} mm")
```

## Task 4: Wettest Month Per County

In this task, I use NumPy's argmax() function with axis=1 to identify the month with the highest rainfall for each county. I then use zip() to pair each county with its corresponding wettest month.

```python
# Find the position of the highest rainfall for each county

wettest_indices = np.argmax(R, axis=1)


# Convert the positions into month names

wettest_months = [months[index] for index in wettest_indices]


# Display the wettest month for each county

print("WETTEST MONTH PER COUNTY")
print("-" * 40)

for county, month in zip(counties, wettest_months):

    print(f"{county:<12}: {month}")
```

## Task 5: Dry Month Analysis

In this task, I use a NumPy boolean mask to identify rainfall readings below 50 mm. I then use NumPy's sum() function along axis=1 to count the number of dry months in each county.

```python
# Create a boolean mask for rainfall below 50 mm

dry_mask = R < 50


# Count the number of dry months for each county

dry_month_counts = np.sum(dry_mask, axis=1)


# Display the results

print("DRY MONTHS PER COUNTY")
print("-" * 40)

for county, count in zip(counties, dry_month_counts):

    print(f"{county:<12}: {count} dry months")
```

## Task 6: Recommended Planting Windows

In this task, I identify the two consecutive months with the highest combined rainfall for each county. I use NumPy's shifted-array technique to calculate the rainfall totals for all consecutive-month pairs, then use argmax() to identify the best planting window for each county.

```python
# Calculate rainfall for consecutive two-month periods

two_month_rainfall = R[:, :-1] + R[:, 1:]


# Find the position of the highest combined rainfall

planting_indices = np.argmax(two_month_rainfall, axis=1)


# Convert the positions into planting window names

planting_windows = [
    f"{months[index]} - {months[index + 1]}"
    for index in planting_indices
]


# Get the highest combined rainfall for each county

highest_window_rainfall = np.max(two_month_rainfall, axis=1)


# Display the recommended planting windows

print("RECOMMENDED PLANTING WINDOWS")
print("-" * 50)

for county, window, rainfall in zip(
    counties, planting_windows, highest_window_rainfall
):

    print(f"{county:<12}: {window:<20} ({rainfall} mm)")
```

## Task 7: Exporting the Rainfall Summary

In this task, I export the final county-level analysis into a CSV file using Python's csv module. The file contains one row for each county and includes annual rainfall totals, wettest months, dry-month counts and recommended planting windows.

```python
# Create the rainfall summary CSV file

with open("rainfall_summary.csv", "w", newline="", encoding="utf-8") as file:

    writer = csv.writer(file)

    # Write the required column headings

    writer.writerow([
        "county",
        "annual_total_mm",
        "wettest_month",
        "dry_months",
        "planting_window"
    ])

    # Write one row for each county

    writer.writerows(zip(
        counties,
        annual_totals.tolist(),
        wettest_months,
        dry_month_counts.tolist(),
        planting_windows
    ))


print("rainfall_summary.csv created successfully!")
```

```python
# Read and verify the generated CSV file

with open("rainfall_summary.csv", "r", encoding="utf-8") as file:

    reader = csv.reader(file)

    csv_data = list(reader)


print("CSV COLUMN HEADINGS:")
print(csv_data[0])

print("\nNUMBER OF COUNTY RECORDS:", len(csv_data) - 1)

print("\nCSV CONTENTS:")

for row in csv_data:
    print(row)
```
